**Description:** This notebook executes the end-to-end Data Cleaning, Data Type Transformation, ETL processes, and Executive Visualizations for the public grievances dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load initial raw governance dataset
df = pd.read_csv('/content/CM_Helpline_Data.csv')
print("--- Raw Dataset Preview ---")
print(df.head())
```

```python
# Inspect structural metadata of the dataframe
df.info()
```

```python
# 1. Define explicit schemas for positional headers
column_names = ['Complaint_ID', 'Citizen_Name', 'District', 'Department',
                'Complaint_Type', 'Registered_Date', 'Resolution_Days',
                'Current_Status', 'Assigned_Level']

# 2. Re-import dataset by enforcing structural schema configuration
df = pd.read_csv('/content/CM_Helpline_Data.csv', header=None, names=column_names)

# 3. Verify post-mapping database structure
df.info()
```

```python
# Parse 'Registered_Date' attribute from object type to standard Datetime format
df['Registered_Date'] = pd.to_datetime(df['Registered_Date'])

# Verify data type conversion vector
df.info()
```

```python
# Audit structural missing values across features
df.isnull().sum()
```

```python
# Impute missing records in 'Resolution_Days' with 0 to indicate immediate or active pendency status
df['Resolution_Days'] = df['Resolution_Days'].fillna(0)

# Re-validate complete clean data schema
df.info()
```

## Task 1: Department-wise Workload & Ticket Analysis
**Objective:** Compute corporate workloads across public departments to identify operational bottlenecks.

```python
# Generate categorical distribution matrix for departments
dept_summary = df['Department'].value_counts()
print("---- Department Total Ticket Distribution ---- ")
print(dept_summary)

# Render executive distribution bar metric visualization
dept_summary.plot(kind='bar', color='teal', edgecolor='black', figsize=(8, 5))
plt.title('CM Helpline: Department-wise Workload Analysis', fontsize=14, fontweight='bold')
plt.xlabel('Government Departments', fontsize=12)
plt.ylabel('Volume of Registered Complaints', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Task 2: Geographic Distribution of Pending Tickets
**Objective:** Isolate pending administrative complaints by district to assist in executive triage and alerts.

```python
# 1. Filter out unresolved public ticket matrices
pending_df = df[df['Current_Status'] == 'Pending']

# 2. Calculate localized ticketing distributions
district_pending = pending_df['District'].value_counts()
print("--- District-wise Pending Caseloads ---")
print(district_pending)

# 3. Render categorical visualization matrix via Pie Chart mapping
district_pending.plot(kind='pie', autopct='%1.1f%%', startangle=90, figsize=(6,4), cmap='Set2')
plt.title('Distribution of Pending Complaints by District', fontsize=14, fontweight='bold')
plt.ylabel('') 
plt.show()
```

## Task 3: Operational Level Bottleneck Identification
**Objective:** Map ticket blocks across SLA operational tiers (L1, L2, L3 Support Nodes).

```python
# Calculate pending ticket vector across systemic organizational hierarchies
level_bottleneck = pending_df['Assigned_Level'].value_counts()
print("--- Structural Backlogs by Operational SLA Level ---")
print(level_bottleneck)
```

## Task 4: Daily Ticketing Trend Analysis
**Objective:** Profile daily input spikes to forecast system volume anomalies over operational timelines.

```python
# 1. Profile chronological influx frequencies
daily_trend = df.groupby('Registered_Date')['Complaint_ID'].count()
print("--- Time Series Ticketing Influx Data ---")
print(daily_trend)

# 2. Generate predictive/historical line vector trend charts
plt.figure(figsize=(10, 5))
daily_trend.plot(kind='line', marker='o', color='crimson', linewidth=2, linestyle='-')

plt.title('CM Helpline: Daily Influx of Complaints Trend Mapped', fontsize=14, fontweight='bold')
plt.xlabel('Registration Date Timeline', fontsize=12)
plt.ylabel('Number of Complaints Received', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

## Task 5: Cross-Tabulation Matrix (District vs. Department Backlogs)

```python
# Generate relational multi-axis pivot matrix for operational visibility
pending_matrix = pd.crosstab(pending_df['District'], pending_df['Department'])
print("---- Cross-Tabulation: Pending Influx Matrix (District x Department) ----")
print(pending_matrix)
```

## Task 6: Export Clean Production Artifacts

```python
# Export engineered metrics data to clean production environment file
df.to_csv('cleaned_governance_data.csv', index=False)
print("Production artifact exported successfully.")
```
